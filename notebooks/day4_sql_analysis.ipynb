```python
import pandas as pd

orders = pd.read_csv("../data/olist_orders_dataset.csv")

order_items = pd.read_csv("../data/olist_order_items_dataset.csv")

reviews = pd.read_csv("../data/olist_order_reviews_dataset.csv")

products = pd.read_csv("../data/olist_products_dataset.csv")

print(orders.shape)
print(order_items.shape)
print(reviews.shape)
print(products.shape)
```

```python
returned_orders = reviews[reviews['review_score'] <= 2]

returned_orders.head()
```

```python
print(returned_orders.shape)
```

```python
returns = returned_orders[['order_id']]

returns.head()
```

```python
merged = returns.merge(
    order_items,
    on='order_id'
)

merged.head()
```

```python
print(merged.shape)
```

```python
merged = merged.merge(
    products,
    on='product_id'
)

merged.head()
```

```python
print(merged.shape)
```

```python
sku_returns = merged.groupby(
    'product_id'
).size().reset_index(
    name='return_count'
)

sku_returns.head()
```

```python
sku_returns = sku_returns.sort_values(
    by='return_count',
    ascending=False
)

sku_returns.head(10)
```

```python
sku_returns.to_csv(
    "../reports/sku_return_rates.csv",
    index=False
)

print("Report saved successfully")
```
